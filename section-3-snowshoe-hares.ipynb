# Data Frame Using a Dictionary

import pandas as pd

data_frame = pd.DataFrame([[i] for i in range(7)], columns =['data'])

# Introducing weeks as dictionary
weeks = {0:'Sunday', 1:'Monday', 2:'Tuesday', 3:'Wednesday', 
4:'Thursday', 5:'Friday', 6:'Saturday'}

# Mapping the dictionary keys to the data frame.
data_frame['new_data_1'] = data_frame['data'].map(weeks)
print (data_frame)

```python
hares['sex'] = hares['sex'].str.lower().str.strip()
# print(sum(hares['sex'].value_counts()))
# print(sum(hares['sex']=='?'))

sex_dict = {'f?':'female',
            'f':'female',
            'pf':'female',
            'm?':'male',
            'm':'male',
            '?':np.nan
            }

hares['sex_simple'] = hares['sex'].map(sex_dict)
# print(sum(hares['sex_simple'].value_counts())+sum(hares['sex']=='?'))
hares['sex_simple'].value_counts(dropna=False)
```
